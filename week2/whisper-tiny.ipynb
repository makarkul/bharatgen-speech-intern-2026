```python
import glob
import json
import librosa
import torch
from transformers import AutoProcessor, AutoModelForSpeechSeq2Seq

model_id = "openai/whisper-tiny"
processor = AutoProcessor.from_pretrained(model_id)
model = AutoModelForSpeechSeq2Seq.from_pretrained(model_id)
model.eval()

print("model loaded")
```

```python
# run on all 5 Week 1 clips
results = {}
for path in sorted(glob.glob("bharatgen-speech-intern-2026/week1/data/audio_clips/*.wav")):
    audio, sr = librosa.load(path, sr=16000)
    inputs = processor(audio, sampling_rate=16000, return_tensors="pt")
    with torch.no_grad():
        ids = model.generate(inputs.input_features)
    results[path] = processor.batch_decode(ids, skip_special_tokens=True)[0]
    print(f"{path}: {results[path]}")
```

```python
# save as Week 3 baseline
with open("whisper_tiny_baseline.json", "w") as f:
    json.dump(results, f, ensure_ascii=False, indent=2)

print("saved to whisper_tiny_baseline.json")
```
