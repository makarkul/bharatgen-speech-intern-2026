### Sentiment Analysis

```python
from transformers import pipeline

classifier = pipeline("sentiment-analysis")
classifier("I've been waiting for a HuggingFace course my whole life.")
```

### Mask Filling
Used when we want to train a model to predict what will come in the blank.

```python
from transformers import pipeline

unmasker = pipeline("fill-mask")
unmasker("This course will teach you all about <mask> languages.", top_k=3)
```

### Named Entry Recognition

```python
import warnings
from transformers import pipeline

# Suppress warnings to clean up output
warnings.filterwarnings("ignore")

# Correcting the argument name to 'aggregation_strategy'
ner = pipeline("ner", model="dbmdz/bert-large-cased-finetuned-conll03-english", aggregation_strategy="simple")
ner("My name is Sylvain and I work at Hugging Face in Brooklyn.")
```
