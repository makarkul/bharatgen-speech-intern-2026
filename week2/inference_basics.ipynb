## AutoTokenizer

AI models have different tokenizers for different LLMs. When we want to run them, we dont have to worry about which model uses which tokenizer method. Autotokenizer takes care of that for us.

```python
from transformers import AutoTokenizer

checkpoint = "distilbert-base-uncased-finetuned-sst-2-english"
tokenizer = AutoTokenizer.from_pretrained(checkpoint)
```

#### Testing with a sentence

```python
tokenizer("I am going to eat pizza!")
```

#### We can also pass a list of sentences, as it will have faster processing than processing sentences differenty

```python
raw_inputs = [
    "I've been waiting for a HuggingFace course my whole life.",
    "I hate this so much!",
]
inputs = tokenizer(raw_inputs, padding=True, truncation=True, return_tensors="pt")
print(inputs)

#padding is used to make all  the inputs the same length, this is useful when we want inputs to be of same length

#truncation is there as a safety net. there is a limit on the length of a sentence, and if limit is surpassed, truncation will chop off ends of the sentence.

#return tensort = "pt": this gives the resultant as a pytorch tensor
```

Here  

101 - This is called CLS token. Marks the beginning of string  

102 - SEP token. This tells the ending of a sentence.  

Attention Mask - This tells us which of the numbers are real and fake. Because we have given padding to the string, it is filled with 0. We need to tell computer that the numbers are just filler, otherwise it will include the 0's also while the models learning.

### AutoModel

```python
from transformers import AutoModel

checkpoint = "distilbert-base-uncased-finetuned-sst-2-english"
model = AutoModel.from_pretrained(checkpoint)
```

#### Checking with list of Inputs we have.

```python
outputs = model(**inputs)                   # ** here is used to pass the dictionary key-value pairs seperately
print(outputs.last_hidden_state.shape)
```
