#### NLP
Natural Language Processing is the process of making computers understand human languages.  
There are various parts of NLP like NER, Sentiment, Machine Translation.  

NER - named entry recognition is used to identify the parts of a sentence. Like if i say Chaitanya is going to the market, the NER will recognise that Chaitanya is a person, and market is a place.  

Sentiment - machines work with numbers and statistics, its hard for them to understand emotions. normally while speaking, we may be angry, happy or sarcastic, that is hard for machines to understand. Here the sentiment part becomes useful.  

Machine Translation - When we want to translate from one language, we use this feature. This feature is said to be useful when we convert a sentence into another language, and when it comes back, it retains its meaning and structure.

### LLM

In the NLP approach, if we wanted to make a model for a specific task, we would have to do it from scratch, and we'd have to make another model for another specific task.  

With LLM, we train one model with a LOT of data, this model gets trained on a lot of general data. Now, it can perform a variety of tasks, without being specifically trained for a task. Basically, it can perform a lot of tasks as we have generalised it.  

LLM is a subset of NLP.

### Transformers
A type of neural network that is made to have a memory and is useful when sequences are involved. When we give it a sentence, it remembers the words that came before, which provides it with a better context of the next upcoming words.

#### 3 Ways to make AI behave the way you want

1. Pre Training  
Train the whole model from scratch  

2. Fine Tuning
Take an already pretrained model, then train it to specialize in a certain tasks. The weights are affected slightly in this.  

3. Prompting
When we give instructions in the chat within the context window. 

#### 3 Types of Transformers

1. Encoder - used to understand text  

2. decoder - used to generate text  

3. encoder-decoder - when we need to convert from one for to another. like in the case of language translation

### How Language models learn

#### Masked Lanugage Model - Bert
BERT - BiDirectional Encoder and Representation for Transformer  

In this model, some words are hidden and the model has to guess them. Kind of like fill in the blanks, the model will guess the blanks. The model trains bidirectionally, because it has to look at data on both sides. When we want to understand some text, we need the future context and then MLM is useful. The masking is important because it forces the model to actually understand the ccontext of the sentence.

#### Causal Language Modeling - GPT
In this, the model predicts what word will come first. It is one directional, so model will only see what came before it.  
This is better for generative texts. 

#### BART
BiDirectional Auto-Regressive Transformer  

When we need understanding and generation both.  
In this, we convert one sequence of text into another. Like in translation of languages.  
The BERT part reads the input and understands it, basically encoder. The GPT part of it writes the output, like decoder.  

Here, auto regressive means it writes one word at a time.
