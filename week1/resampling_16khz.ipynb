## Resampling Audio file to 16kHz and Comparing



```python
import torch
import torchaudio
from IPython.display import Audio
import matplotlib.pyplot as plt
```

```python
# Load the 48 kHz version
waveform_48k, sr_48k = torchaudio.load(
    "data/resampling_demo/eng_sample_48k.wav"
)

print("Shape:", waveform_48k.shape)
print("Sample rate:", sr_48k)
print("Duration:", waveform_48k.shape[1] / sr_48k, "seconds")
```

```python
# Resample 48 kHz → 16 kHz
resampler = torchaudio.transforms.Resample(
    orig_freq=48000,
    new_freq=16000
)

waveform_16k = resampler(waveform_48k)

print("48 kHz shape:", waveform_48k.shape)
print("16 kHz shape:", waveform_16k.shape)
print("Ratio:", waveform_48k.shape[1] / waveform_16k.shape[1])
```

```python
# Listen to the 48 kHz original
print("48 kHz version:")
Audio(waveform_48k.numpy(), rate=48000)
```

```python
# Listen to the 16 kHz resampled version
print("16 kHz version:")
Audio(waveform_16k.numpy(), rate=16000)
```

### Differences

#### 48kHz
The audio definitely sounds more richer and vibrant. You can hear everything well including the background noises.

#### 16kHz
The audio sounds kind of muffled, like it has been surpressed(which it literally is).  
The background noise also has reduced quiet a bit.  
The audio overall sounds a little hazy.  

But there is no problem in differentiting the speech. There is no confusion in the words that are being said.  

So for speech recognition, it is a win. Because speech almost sounds the same. But it occupies less space and it can be processed much faster.  

So for ASR models, it should cause no error when we downsize from 48kHz audio to 16kHz. It will infact improve the efficiency, because of the faster processing rate.
