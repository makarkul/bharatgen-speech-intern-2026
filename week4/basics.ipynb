## Embedding

When we need to feed something to the computer, we need numbers. We cant just feed an image or audio file. We need a way to turn them into numbers. 

Embedding is the solution to represent this. 
EMbedding is a vector, basically a list of numbers. The number of numbers decides the dimension of the vector. If 3 numbers then 3D. If 100, then 100D.

If two items are similar/related, then they are likely to have similar vectors.

### Testing with Embedding

```python
from sentence_transformers import SentenceTransformer 
from sklearn.metrics.pairwise import cosine_similarity 

# 1. Load a pretrained embedding model
model = SentenceTransformer('all-MiniLM-L6-v2')

# 2. Words we want to embed
words = ["cat", "dog", "banana"]

# 3. Convert each word into a vector
embeddings = model.encode(words)

# 4. Check the shape
print("Shape:", embeddings.shape)
# Output: (3, 384)  -> 3 words, each turned into a 384-number vector

# 5. Compare them using cosine similarity
print("cat vs dog   :", cosine_similarity([embeddings[0]], [embeddings[1]])[0][0])
print("cat vs banana:", cosine_similarity([embeddings[0]], [embeddings[2]])[0][0])
```

We can see cat and dog have much more similarity than cat and banana.
